```python
import yfinance as yf
import pandas as pd

# Test possible Yahoo Finance symbols
test_symbols = [
    "AIRTEL.MW",
    "TNM.MW",
    "AIRTEL",
    "TNM"
]

for symbol in test_symbols:
    try:
        data = yf.download(
            symbol,
            period="5y",
            auto_adjust=False,
            progress=False
        )

        if not data.empty:
            print(f"\n{symbol}: DATA FOUND")
            print(data.tail(3))
        else:
            print(f"{symbol}: No data returned")

    except Exception as e:
        print(f"{symbol}: Error - {e}")
```
